### Q.12 - Quel est le nombre minimal de bits nécessaire pour coder un entier sans signe, compris entre 0 et 255 ?

Pour coder un entier sans signe compris entre **0 et 255**, il faut pouvoir représenter **256 valeurs différentes**.

On cherche donc le plus petit entier `n` tel que :

``2^n ≥ 256``

En testant :

- `2^7 = 128` → insuffisant  
- `2^8 = 256` → correspond exactement

✅ **Réponse :** Il faut **8 bits** pour coder un entier entre 0 et 255.

### Q.14 - Fonction `histogramme(M)` : Calcul de l'histogramme d'une image couleur

```python
import numpy as np
def histogramme(M):
    # Initialisation d'une matrice H de 3 lignes et 256 colonnes remplie de zéros
    # H = [[0] * 256 for _ in range(3)]
    H=np.zeros((256,3))
    # Parcours de la matrice M
    for ligne in M:
        for pixel in ligne:
            r, v, b = pixel  # Extraction des composantes rouge, vert, bleu
            H[0][r] += 1     # Histogramme rouge
            H[1][v] += 1     # Histogramme vert
            H[2][b] += 1     # Histogramme bleu

    return H
```

```python
from PIL import Image
import numpy as np

def image_to_matrix(image_path):
    # Ouvrir l'image
    image = Image.open(image_path)
    
    # Convertir l'image en un tableau NumPy
    image_matrix = np.array(image)
    
    return image_matrix

# Exemple d'utilisation
image_path = 'image.png'
matrix = image_to_matrix(image_path)
print(matrix)
```
